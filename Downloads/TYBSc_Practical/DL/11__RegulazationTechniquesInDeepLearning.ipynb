```python
# ============================================================
# PRACTICAL: REGULARIZATION TECHNIQUES IN DEEP LEARNING
# DATASET: FASHION MNIST
# ============================================================


# -------------------------------
# 1. IMPORT REQUIRED LIBRARIES
# -------------------------------

# Import TensorFlow for deep learning.
import tensorflow as tf

# Import the Sequential model for creating neural networks.
from tensorflow.keras.models import Sequential

# Import layers required to build the neural network.
from tensorflow.keras.layers import Flatten, Dense, Dropout

# Import L2 regularization.
from tensorflow.keras.regularizers import l2

# Import EarlyStopping callback.
from tensorflow.keras.callbacks import EarlyStopping

# Import Matplotlib for plotting graphs.
import matplotlib.pyplot as plt


# -------------------------------
# 2. LOAD THE FASHION MNIST DATASET
# -------------------------------

# Load Fashion MNIST training and testing data.
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()


# -------------------------------
# 3. NORMALIZE THE DATA
# -------------------------------

# Convert pixel values from integers to floating-point numbers.
x_train = x_train.astype("float32")

# Convert testing pixel values to floating-point numbers.
x_test = x_test.astype("float32")

# Scale training pixel values from 0-255 to 0-1.
x_train = x_train / 255.0

# Scale testing pixel values from 0-255 to 0-1.
x_test = x_test / 255.0


# -------------------------------
# 4. CREATE A BASELINE MODEL
# -------------------------------

# Create a sequential neural network without regularization.
baseline_model = Sequential([

    # Convert each 28x28 image into a one-dimensional vector.
    Flatten(input_shape=(28, 28)),

    # Add a fully connected hidden layer with 128 neurons.
    Dense(128, activation="relu"),

    # Add a second hidden layer with 64 neurons.
    Dense(64, activation="relu"),

    # Add an output layer with 10 neurons for 10 classes.
    Dense(10, activation="softmax")
])


# -------------------------------
# 5. COMPILE THE BASELINE MODEL
# -------------------------------

# Configure the model for training.
baseline_model.compile(

    # Use Adam optimizer to update model weights.
    optimizer="adam",

    # Use sparse categorical crossentropy for integer labels.
    loss="sparse_categorical_crossentropy",

    # Track accuracy during training.
    metrics=["accuracy"]
)


# -------------------------------
# 6. TRAIN THE BASELINE MODEL
# -------------------------------

# Train the baseline model for 15 epochs.
baseline_history = baseline_model.fit(

    # Provide training images to the model.
    x_train,

    # Provide correct training labels.
    y_train,

    # Use 20% of training data for validation.
    validation_split=0.2,

    # Train the model for 15 complete passes.
    epochs=15,

    # Use 64 images in each training batch.
    batch_size=64,

    # Display training progress.
    verbose=1
)


# -------------------------------
# 7. CREATE A REGULARIZED MODEL
# -------------------------------

# Create a sequential neural network with regularization.
regularized_model = Sequential([

    # Convert each 28x28 image into a one-dimensional vector.
    Flatten(input_shape=(28, 28)),

    # Add a dense layer with L2 weight regularization.
    Dense(128, activation="relu", kernel_regularizer=l2(0.001)),

    # Randomly deactivate 30% of neurons during training.
    Dropout(0.3),

    # Add a second dense layer with L2 regularization.
    Dense(64, activation="relu", kernel_regularizer=l2(0.001)),

    # Randomly deactivate 30% of neurons during training.
    Dropout(0.3),

    # Add the output layer for 10 clothing classes.
    Dense(10, activation="softmax")
])


# -------------------------------
# 8. COMPILE THE REGULARIZED MODEL
# -------------------------------

# Configure the regularized model for training.
regularized_model.compile(

    # Use Adam optimizer.
    optimizer="adam",

    # Use sparse categorical crossentropy loss.
    loss="sparse_categorical_crossentropy",

    # Track accuracy during training.
    metrics=["accuracy"]
)


# -------------------------------
# 9. CREATE EARLY STOPPING
# -------------------------------

# Stop training when validation loss stops improving.
early_stopping = EarlyStopping(

    # Monitor the validation loss.
    monitor="val_loss",

    # Wait for 2 epochs before stopping.
    patience=2,

    # Restore the weights from the best validation epoch.
    restore_best_weights=True
)


# -------------------------------
# 10. TRAIN THE REGULARIZED MODEL
# -------------------------------

# Train the regularized model.
regularized_history = regularized_model.fit(

    # Provide training images.
    x_train,

    # Provide training labels.
    y_train,

    # Use 20% of training data for validation.
    validation_split=0.2,

    # Set the maximum number of epochs.
    epochs=15,

    # Use 64 images per batch.
    batch_size=64,

    # Apply early stopping during training.
    callbacks=[early_stopping],

    # Display training progress.
    verbose=1
)


# -------------------------------
# 11. EVALUATE BOTH MODELS
# -------------------------------

# Evaluate the baseline model using testing data.
baseline_loss, baseline_accuracy = baseline_model.evaluate(

    # Provide testing images.
    x_test,

    # Provide testing labels.
    y_test,

    # Hide evaluation progress.
    verbose=0
)

# Evaluate the regularized model using testing data.
regularized_loss, regularized_accuracy = regularized_model.evaluate(

    # Provide testing images.
    x_test,

    # Provide testing labels.
    y_test,

    # Hide evaluation progress.
    verbose=0
)


# -------------------------------
# 12. DISPLAY ACCURACY RESULTS
# -------------------------------

# Print the baseline model testing accuracy.
print("Baseline Test Accuracy:",
      round(baseline_accuracy * 100, 2), "%")

# Print the regularized model testing accuracy.
print("Regularized Test Accuracy:",
      round(regularized_accuracy * 100, 2), "%")


# -------------------------------
# 13. PLOT TRAINING AND VALIDATION LOSS
# -------------------------------

# Create a figure for comparing loss.
plt.figure(figsize=(10, 5))

# Plot baseline training loss.
plt.plot(baseline_history.history["loss"], label="Baseline Training Loss")

# Plot baseline validation loss.
plt.plot(baseline_history.history["val_loss"], label="Baseline Validation Loss")

# Plot regularized training loss.
plt.plot(regularized_history.history["loss"], label="Regularized Training Loss")

# Plot regularized validation loss.
plt.plot(regularized_history.history["val_loss"], label="Regularized Validation Loss")

# Add a title to the graph.
plt.title("Training and Validation Loss")

# Add a label to the horizontal axis.
plt.xlabel("Epochs")

# Add a label to the vertical axis.
plt.ylabel("Loss")

# Display the graph legend.
plt.legend()

# Display the graph.
plt.show()


# -------------------------------
# 14. PLOT TRAINING AND VALIDATION ACCURACY
# -------------------------------

# Create a figure for comparing accuracy.
plt.figure(figsize=(10, 5))

# Plot baseline training accuracy.
plt.plot(baseline_history.history["accuracy"], label="Baseline Training Accuracy")

# Plot baseline validation accuracy.
plt.plot(baseline_history.history["val_accuracy"], label="Baseline Validation Accuracy")

# Plot regularized training accuracy.
plt.plot(regularized_history.history["accuracy"], label="Regularized Training Accuracy")

# Plot regularized validation accuracy.
plt.plot(regularized_history.history["val_accuracy"], label="Regularized Validation Accuracy")

# Add a title to the graph.
plt.title("Training and Validation Accuracy")

# Add a label to the horizontal axis.
plt.xlabel("Epochs")

# Add a label to the vertical axis.
plt.ylabel("Accuracy")

# Display the graph legend.
plt.legend()

# Display the graph.
plt.show()


# -------------------------------
# 15. DISPLAY TRAINING INFORMATION
# -------------------------------

# Print the number of epochs completed by the regularized model.
print("Regularized Model Epochs Completed:",
      len(regularized_history.history["loss"]))
```
