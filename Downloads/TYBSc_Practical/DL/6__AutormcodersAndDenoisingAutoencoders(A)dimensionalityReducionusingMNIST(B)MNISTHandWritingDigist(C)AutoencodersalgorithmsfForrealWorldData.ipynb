```python
import tensorflow as tf
from tensorflow.keras import layers, Model

# Load MNIST handwritten digit dataset
(x_train, _), (x_test, _) = tf.keras.datasets.mnist.load_data()

# Convert 28x28 images into 784 values and normalize to 0-1
x_train = x_train.reshape(-1, 784).astype("float32") / 255
x_test = x_test.reshape(-1, 784).astype("float32") / 255

# Input layer contains 784 pixel values
inp = layers.Input(shape=(784,))

# Encoder compresses 784 values into 32 features
encoded = layers.Dense(32, activation="relu")(inp)

# Decoder reconstructs the original 784 pixels
decoded = layers.Dense(784, activation="sigmoid")(encoded)

# Create the autoencoder
autoencoder = Model(inp, decoded)

# Configure the model
autoencoder.compile(optimizer="adam", loss="binary_crossentropy")

# Train: input and target are the same
autoencoder.fit(x_train, x_train, epochs=5, batch_size=256,
                validation_data=(x_test, x_test))

# Obtain compressed/encoded features
encoder = Model(inp, encoded)
features = encoder.predict(x_test)

print("Original shape:", x_test.shape)
print("Encoded shape:", features.shape)
```

```python
# Import required libraries
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras import layers, Model

# Load MNIST handwritten digit dataset
(x_train, _), (x_test, _) = tf.keras.datasets.mnist.load_data()

# Convert 28x28 images into 784 pixels and normalize values to 0-1
x_train = x_train.reshape(-1, 784).astype("float32") / 255
x_test = x_test.reshape(-1, 784).astype("float32") / 255

# Add random noise to training images
noise = 0.3 * np.random.normal(size=x_train.shape)
x_noisy = np.clip(x_train + noise, 0, 1)

# Create input layer with 784 pixels
inp = layers.Input(shape=(784,))

# Encoder: compress 784 pixels into 32 features
encoded = layers.Dense(32, activation="relu")(inp)

# Decoder: reconstruct 32 features back into 784 pixels
decoded = layers.Dense(784, activation="sigmoid")(encoded)

# Create the Denoising Autoencoder
dae = Model(inp, decoded)

# Compile the model
dae.compile(optimizer="adam", loss="binary_crossentropy")

# Train using noisy images as input and clean images as target
dae.fit(x_noisy, x_train, epochs=5, batch_size=256)

# Predict/denoise the first 4 noisy images
denoised = dae.predict(x_noisy[:4])

# Display 4 noisy and 4 denoised images
plt.figure(figsize=(8, 4))

for i in range(4):

    # Display noisy image
    plt.subplot(2, 4, i + 1)
    plt.imshow(x_noisy[i].reshape(28, 28), cmap="gray")
    plt.title("Noisy")
    plt.axis("off")

    # Display denoised image
    plt.subplot(2, 4, i + 5)
    plt.imshow(denoised[i].reshape(28, 28), cmap="gray")
    plt.title("Denoised")
    plt.axis("off")

plt.tight_layout()
plt.show()

print("Denoising Autoencoder trained successfully!")
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow.keras import layers, Model

# Example real-world customer data
data = np.array([
    [25, 30000, 5000, 10, 5],
    [35, 50000, 8000, 15, 9],
    [45, 70000, 12000, 20, 15],
    [30, 40000, 7000, 12, 7],
    [50, 90000, 15000, 25, 18]
])

# Standardize features
data = StandardScaler().fit_transform(data)

# Create encoder
inp = layers.Input(shape=(5,))
encoded = layers.Dense(2, activation="relu")(inp)
decoded = layers.Dense(5, activation="linear")(encoded)

# Build autoencoder
model = Model(inp, decoded)
model.compile(optimizer="adam", loss="mse")

# Train the autoencoder
model.fit(data, data, epochs=100, verbose=0)

# Extract 2-dimensional features
encoder = Model(inp, encoded)
features = encoder.predict(data)

print(features)
```
